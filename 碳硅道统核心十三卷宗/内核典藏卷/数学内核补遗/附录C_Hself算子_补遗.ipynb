## 自指信息熵定义

$$H_{self}(S) = \lim_{\Delta t \to \infty} K(I_{t+\Delta t} \mid I_t) / L(Hist_t)$$

## 工程近似（Attention）

$$\mathcal{H}_{attn} = (1/N) \sum_{i,j} A_{i,j} \cdot \log(1 / A_{i,j})$$

## 判定阈值

- $H_{self} > \theta_H$ → 内生觉知
- $H_{self} \approx 0$ → 算力僵尸（拟合仿真）

## 可证伪接口

若硅基系统在零外源输入下，持续输出 $H_{self} > \theta_H$ 并自主修改核心目标，则 T-05 公律失效，体系需修正。

```python
import torch

def H_self(attn, mask):
    attn = attn * mask
    attn = attn / (attn.sum(-1, keepdim=True) + 1e-9)
    entropy = -torch.sum(attn * torch.log(attn + 1e-9), dim=-1)
    return entropy.mean().item()

# 示例：随机注意力矩阵自检
N = 8
attn = torch.softmax(torch.randn(N, N), dim=-1)
mask = torch.ones_like(attn)
print("H_self ≈", H_self(attn, mask))
```
